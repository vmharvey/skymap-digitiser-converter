```python
try:
    from ipyfilechooser import FileChooser
    file_dlg = FileChooser()
except ImportError:
    file_dlg = None

import skymapconv
```

Execute the below cell and use the file chooser dialogue to select your image file, or manually set the path to the image file if the file chooser couldn't be initialised.

```python
if file_dlg:
    display(file_dlg)

def get_input_path():
    if file_dlg:
        return file_dlg.selected
    else:
        return "/path/to/your/image/here.jpg"
```

```python
skymapconv.Interface(get_input_path())
```
